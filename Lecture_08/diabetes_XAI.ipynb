```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
import shap
```

```python
df = pd.read_csv("diabetes.csv")
df.head()
```

```python
X = df.drop(columns=['Outcome'])
y = df['Outcome']
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

## Yêu cầu 1: 
#### Sử dụng SHAP LinearExplainer để giải thích kết quả từ mô hình Logistic Regression sau

```python
lg = LogisticRegression(max_iter = 500)
lg.fit(X_train, y_train)
lg.score(X_train, y_train), lg.score(X_test, y_test)
```

## Yêu cầu 2

Sử dụng TreeShap để giải thích kết quả từ các mô hình dạng cây như Decision Tree, Random Forest, Adaboost

```python
rfc = RandomForestClassifier(max_depth = 5)
rfc.fit(X_train, y_train)
rfc.score(X_train, y_train), rfc.score(X_test, y_test)
```

## Yêu cầu 3

Sử dụng SHAP kernel để giải thích kết quả từ các cây như Naive Bayes, KNN, MLP. 

```python
mlp = MLPClassifier()
mlp.fit(X_train, y_train)
mlp.score(X_train, y_train), mlp.score(X_test, y_test)
```
